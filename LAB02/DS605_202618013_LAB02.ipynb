## Task-2

```python
import numpy as np
np.random.seed(42)

M1 = np.random.randint(1, 10, size=(3, 3))
M2 = np.random.randint(1, 10, size=(3, 3))
print("Matrix M1: ", M1)
print("Matrix M2: ", M2)

addition = M1 + M2
print("ADDITION (M1 + M2)", addition)

elementwise_mult = M1 * M2
print("ELEMENT-WISE MULTIPLICATION (M1 * M2)")
print(elementwise_mult)

matmul_at = M1 @ M2
matmul_fn = np.matmul(M1, M2)
print("MATRIX MULTIPLICATION")
print("Using @ operator: ", matmul_at)
print("\nUsing np.matmul(): ", matmul_fn)
print(f"\nBoth methods equal? {np.array_equal(matmul_at, matmul_fn)}")

transpose_M1 = M1.T
det_M1 = np.linalg.det(M1)
print("TRANSPOSE AND DETERMINANT OF M1")
print("\nTranspose of M1 (M1.T): ", transpose_M1)
print(f"\nDeterminant of M1: {det_M1:.4f}")
print("INVERSE AND VERIFICATION")

if not np.isclose(det_M1, 0):
    inv_M1 = np.linalg.inv(M1)
    print("Inverse of M1:", inv_M1)

    identity_check = M1 @ inv_M1
    print("\nM1 @ inv(M1):", identity_check)

    is_identity = np.allclose(identity_check, np.eye(M1.shape[0]))
    print(f"\nDoes M1 @ inv(M1) match the identity matrix? {is_identity}")
else:
    print(f"M1 has determinant ~0 ({det_M1}) implies matrix is singular, "
          f"no inverse exists.")
```

## Task-3

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
mu = 50       # chosen population mean
sigma = 10    # chosen population standard deviation
n_samples = 2000

print("CHOSEN DISTRIBUTION PARAMETERS")
print(f"Population mean (mu): {mu}")
print(f"Population std dev (sigma): {sigma}")
print(f"Number of samples: {n_samples}")

data = np.random.normal(loc=mu, scale=sigma, size=n_samples)
print("\n SAMPLE DATA (first 10 values)")
print(data[:10])

sample_mean = np.mean(data)
sample_std = np.std(data)
print("\n SAMPLE STATISTICS vs CHOSEN PARAMETERS")
print(f"Sample mean: {sample_mean:.4f}  (chosen mu   = {mu})")
print(f"Sample std dev: {sample_std:.4f}  (chosen sigma = {sigma})")
print(f"Mean difference: {abs(sample_mean - mu):.4f}")
print(f"Std dev difference: {abs(sample_std - sigma):.4f}")
print("""\n  Observation: With a large sample size (2000), the sample mean and
sample standard deviation are very close to the chosen population parameters
(mu=50, sigma=10), consistent with the Law of Large Numbers - as sample size
grows, sample statistics converge to the true population parameters. """)

# 4. Plot histogram
plt.figure(figsize=(9, 6))
counts, bins, patches = plt.hist(
    data, bins=40, color="#4C72B0", edgecolor="black",
    alpha=0.75, density=True, label="Sample histogram"
)

# Overlay the theoretical normal distribution curve for comparison
x = np.linspace(data.min(), data.max(), 200)
theoretical_pdf = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(
    -0.5 * ((x - mu) / sigma) ** 2
)
plt.plot(x, theoretical_pdf, color="darkred", linewidth=2, label=f"Theoretical N({mu}, {sigma}²) PDF")
plt.axvline(sample_mean, color="black", linestyle="--", linewidth=1.5, label=f"Sample mean = {sample_mean:.2f}")

plt.title(f"Histogram of {n_samples} Samples from Normal Distribution\n")
plt.xlabel("Value")
plt.ylabel("Density")
plt.legend()
plt.grid(alpha=0.3)
```

```python
df = pd.read_csv("train.csv")
print("\n" ,df.head())  #head
print("\n", df.tail())  #tail
print("\n Shape:", df.shape)  #shape
print("\n Columns:", df.columns)  #columns
print("\n\n")
df.info()  #info
print("\n\n")
df.describe()  #descibe

print(df.loc[0:4, ["Name", "Age", "Sex"]])  #loc
print("\n")
print(df.iloc[0:5, 0:3] ) #iloc
print("difference: \n\t loc  → label-based \n\t iloc → integer-position-based")
```

## Task 5 - Filtering and Querying

```python
# Male passengers older than 50: how many?
male_over_50 = df[ (df["Sex"] == "male") & (df["Age"] > 50)]
print("Male passengers older than 50:", len(male_over_50))

# Female first-class passengers: how many, and what percentage survived?
female_first_class = df[ (df["Sex"] == "female") & (df["Pclass"] == 1)]
count_female_first_class = len(female_first_class)
survival_percentage = female_first_class["Survived"].mean() * 100
print("Female first-class passengers:", count_female_first_class)
print("Survival percentage:", survival_percentage)

# Age 20-40, Fare above overall median, and survived: how many?
fare_median = df["Fare"].median()
condition = ( (df["Age"] >= 20) & (df["Age"] <= 40) & (df["Fare"] > fare_median) & (df["Survived"] == 1))
result = df[condition]
print("Number of passengers:", len(result))

# Travelling alone (SibSp=0, Parch=0), age < 30, and did not survive: how many?
not_survived = df[
    (df["SibSp"] == 0) & (df["Parch"] == 0) & (df["Age"] < 30) & (df["Survived"] == 0) ]

print( "Travelling alone, age < 30, did not survive:", len(not_survived) )

# Embarked='S', Pclass 2 or 3, and Fare above Southampton median: how many?
southampton_median = df.loc[
    df["Embarked"] == "S",
    "Fare"
].median()
result = df[ (df["Embarked"] == "S") & (df["Pclass"].isin([2, 3])) & (df["Fare"] > southampton_median) ]
print("Number of passengers:", len(result))

```

## Task 6 - groupby() and Aggregation

```python
# Survival rate by Sex.
survival_by_sex = df.groupby("Sex")["Survived"].mean()
print(survival_by_sex)
print(survival_by_sex * 100)

# Survival rate by Pclass.
print("\n")
survival_by_class = df.groupby("Pclass")["Survived"].mean()
print(survival_by_class)
print(survival_by_class * 100)

# Average Age and Fare by Pclass.
print("\n")
avg_age_fare = df.groupby("Pclass")[["Age", "Fare"]].mean()
print(avg_age_fare)

# Passenger count and survival rate by Sex-Pclass
print("\n")
sex_class_summary = df.groupby(
    ["Sex", "Pclass"] ).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean") )
print(sex_class_summary)
sex_class_summary["Survival_Rate"] *= 100
print(sex_class_summary)

# Passenger count, average Fare, and survival rate by Embarke
print("\n")
embarked_summary = df.groupby("Embarked").agg(
    Passenger_Count=("PassengerId", "count"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean") )
embarked_summary["Survival_Rate"] *= 100
print(embarked_summary)
```

## Task 7 - Missing Values and Fare Outliers

```python
# Find the missing-value count and percentage for every column. Plot missing counts as a bar chart.
missing_count = df.isnull().sum()
print("Missing values: \n", missing_count)

missing_percentage = (
    df.isnull().sum() / len(df)
) * 100
print("\n" ,missing_percentage,"\n")
missing_summary = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percentage": missing_percentage
})
print(missing_summary)

plt.figure(figsize=(10, 5))
df.isnull().sum().plot(kind="bar")
plt.title("Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)
plt.show()

# Fill missing Age values with the mean Age. Show missing counts before and after.
df_mean = df.copy()
print("Missing Age before:", df_mean["Age"].isnull().sum())
age_mean = df_mean["Age"].mean()
print("Mean Age:", age_mean)
df_mean["Age"] = df_mean["Age"].fillna(age_mean)
print("Missing Age after:", df_mean["Age"].isnull().sum())

# Try to fill NaN values by imputing using: mean, median, mode, random value
df_mean = df.copy()  #imputing using mean
df_mean["Age"] = df_mean["Age"].fillna(
    df_mean["Age"].mean() )

df_median = df.copy()  #imputing using median
df_median["Age"] = df_median["Age"].fillna(
    df_median["Age"].median() )

df_mode = df.copy()  #imputing using mode
age_mode = df_mode["Age"].mode()[0]
df_mode["Age"] = df_mode["Age"].fillna(age_mode)

df_random = df.copy()  #imputing using random
np.random.seed(42)
missing_age_count = df_random["Age"].isnull().sum()
random_values = np.random.choice(
    df_random["Age"].dropna(),
    size=missing_age_count,
    replace=True )
df_random.loc[df_random["Age"].isnull(), "Age" ] = random_values

print("Missing Age after random imputation:",  #check if null values are filled
      df_random["Age"].isnull().sum())

# For Fare, calculate Q1, Q3, IQR, 1.5xIQR bounds, and count the outliers
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
fare_outliers = df[ (df["Fare"] < lower_bound) | (df["Fare"] > upper_bound) ]
print("Number of Fare outliers:", len(fare_outliers))
```

## Task 8 - Features and Pivot Table

```python
# Create FamilySize = SibSp + Parch + 1 and IsAlone = 1 when FamilySize = 1
df["FamilySize"] = (df["SibSp"] + df["Parch"] + 1)
# df[["SibSp", "Parch", "FamilySize"]].head()

df["IsAlone"] = np.where(df["FamilySize"] == 1, 1, 0)
df[["FamilySize", "IsAlone"]].head()

# Create a pivot table
pivot_table = pd.pivot_table(
    df,
    index="Sex",
    columns="Pclass",
    values="Survived",
    aggfunc="mean" )
print(pivot_table)

highest_group = pivot_table.stack().idxmax()
highest_value = pivot_table.stack().max()
print("Highest survival group:", highest_group)
print("Highest survival rate:", highest_value)

lowest_group = pivot_table.stack().idxmin()
lowest_value = pivot_table.stack().min()
print("Lowest survival group:", lowest_group)
print("Lowest survival rate:", lowest_value)
```
